```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
a = np.array([3.1,1,-0.5,-3.2,6])
b = np.array([1,3,2.2,5.1,1])

#Multiplicacion escalar
c= a@b
print(f'La multiplicacion escalar de los vectores es: {c} \n')

#Multiplicacion punto a punto
d = a*b
print(f'La multiplicacion punto a punto de los vectores es: {d}')
```

```python
#Calculo de la transpuesta de una matriz A
A = np.array([[2,-1,-3],[4,1.5,-2.5],[7.3,-0.9,0.2]])
print(f'La matriz A corresponde a: \n {A} \n')

#Transpuesta de A
AT = A.transpose()
print(f'La transpuesta de la matriz A corresponde a: \n {AT}')
```

ones = La función numpy.ones() crea un nuevo array (matriz o vector) de la forma y tipo especificados, inicializando todos los elementos en 1 .

round = La funcion numpy.round() Redondea al numero pedido de decimales (round(a,decimals = 0, out=None)).

ceil = La funcion np.ceil(X), redondea hacia arriba al entero mas cercano

floor= La funcion np.floor(x) redondea hacia abajo al entero mas cercano

```python
#Para el ejemplo con ones se crea una matriz de 2 filas y 3 columnas con todos sus elementos en 1
M_Ones = np.ones((2,3))
print(f'La matriz resultante usando np.ones es: \n {M_Ones}\n')

#Para el ejemplo con round se dividira la matriz A por 1.3 y se le pedira que cada uno de sus elementos posea al menos 3 cifras decimales
div = A/1.3
M_round = np.round(div,3)
print(f'La matriz resultante usando np.round es:\n {M_round}\n')

#Para el ejemplo con ceil, se aproximaran todos los valores de la matriz A al entero mas cercano hacia arriba
M_ceil = np.ceil(A)
print(f'La matriz resultante usando np.ceil es: \n {M_ceil}\n')

#Para el ejemplo con floor, se aproximaran todos los valores de la matriz A al entero mas cercano hacia abajo
M_floor = np.floor(A)
print(f'La matriz resultante usando np.ceil es: \n {M_floor}\n')
```

```python
#Obtenga la segunda fila de dicha matriz, imprímalo en consola.
Fila = A[1]
print(f'El valor que corresponde a la segunda fila de A es: {Fila}')
```

```python
#Construccion de funcion senoidal y(n)
n = np.arange(0,100,0.01)
y = np.sin(np.pi*0.12*n)
plt.figure(figsize=(6,4))
plt.plot(n,y,'b')
plt.title(r'Señal Y', fontsize=12)
plt.xlabel('n')
plt.ylabel('Amplitud')
```

```python
#Construccion de funcion senoidal y2(n)
n = np.arange(0,100,0.01)
y2 = np.cos(2*np.pi*0.03*n)
plt.figure(figsize=(6,4))
plt.plot(n,y2,'r')
plt.title(r'Señal Y2', fontsize=12)
plt.xlabel('n')
plt.ylabel('Amplitud')
```

```python
#Genere una tercera señal que sea la suma de estas dos señales, es decir 𝑠[𝑛]=y[n]+𝑦2[𝑛] y una cuarta, que sea el producto de estas, 𝑡[𝑛]=y[n].𝑦2[𝑛].

suma = y+y2
fig_suma = plt.figure(figsize=(6,4))
plt.plot(n,suma,'green')
plt.title(r'Suma de señales', fontsize=12)
plt.xlabel('n')
plt.ylabel('Amplitud')

mult = y*y2
fig_suma = plt.figure(figsize=(6,4))
plt.plot(n,mult,'gray')
plt.title(r'Multiplicación de señales', fontsize=12)
plt.xlabel('n')
plt.ylabel('Amplitud')
```

```python
#Graficar en la misma figura las señales y[n] y y2[n], utilice leyenda y colores diferentes para cada señal. Asigne título a cada eje.

fig, ax = plt.subplots(figsize=(10, 4))

ax.plot(n, y, color='blue', label='y[n] = sin(π·0.12n)')
ax.plot(n, y2, color='red', label='y2[n] = cos(2π·0.03n)')

ax.set_title('Señales senoidales y[n] y y2[n]')
ax.set_xlabel('n')
ax.set_ylabel('Amplitud')
ax.legend()
ax.grid(True)

plt.tight_layout()
plt.show()
```

```python
#Graficar en la misma figura las señales s[n] y t[n], utilice leyenda y colores diferentes para cada señal. Asigne título a cada eje.
fig, ax = plt.subplots(figsize=(10, 4))

ax.plot(n, suma, color='green', label='s[n] = y[n]+y2[n]')
ax.plot(n, mult, color='purple', label='t[n] = y[n]*y2[n]')

ax.set_title('Suma y multiplicación de señales y[n] y y2[n]')
ax.set_xlabel('n')
ax.set_ylabel('Amplitud')
ax.legend(loc='upper right')
ax.grid(True)

plt.tight_layout()
plt.show()
```

```python
#1) Escribir una función que reciba un diccionario con las notas de los alumnos de un curso y devuelva una serie con la nota mínima, la máxima, media y la desviación típica notas
import numpy as np
import random
nombres = ["Sebastian", "Mariana", "Carolina", "Jose", "Felipe", "Johana"]
def diccionario(clave, minimo, maximo):
  if not clave or not isinstance(clave, (list, tuple)):
        raise ValueError("Las claves deben ser una lista o tupla.")
  if not isinstance(minimo, (int, float)) or not isinstance(maximo, (int, float)):
        raise ValueError("El mínimo y máximo deben ser números.")
  if minimo > maximo:
        raise ValueError("El valor mínimo no puede ser mayor que el máximo.")
  lista_notas = {c: [random.randint(minimo, maximo) for i in range(3)] for c in clave}
  promedios = {estudiante: sum(n) / len(n) for estudiante, n in lista_notas.items()}
  return lista_notas, promedios

notas, promedios = diccionario(nombres, 0, 5)
print(f'Las notas registradas son:{notas}')
print(f'El promedio de cada estudiante:{promedios}')

def series(diccionario):
  valores = list(promedios.values())
  return {"Mínimo": round(float(np.min(valores)), 2),"Máximo": round(float(np.max(valores)), 2), "Media": round(float(np.mean(valores)), 2), "Desviación estandar": round(float(np.std(valores)),2)}
notas, promedios = diccionario(nombres, 0, 5)
estadisticas = series(promedios)
minimo, maximo, media, desviacion = estadisticas.values()
print(f'Las estadísticas son: Mínimo de {minimo}, máximo de {maximo}, media de {media} y desviación estandar de {desviacion}')
```

```python
#Para el calculo del indice de masa corporal, se puede usar .columns para conocer el nombre de cada una de las columnas que hacen parte del dataset, sera necesario extraer los datos: weight y height
dpv.columns
```

```python
#Esta funcion cambia el nombre de la columna peso por altura y el de la columna altura por peso, inplace quiere decir que no haya espacios
dpv.rename(columns={'weight':'height','height':'weight'},inplace = True)
dpv.head()
```

```python
#Se debe crear una nueva instancia en el dataset que sera guardada como una columna, esta recibira el nombre de BMI y recibira los datos de peso y altura de cada uno de los pacientes dandole a cada uno su IMC
dpv["BMI"] = dpv["weight"] / ( (dpv["height"] / 100) ** 2 )
```

```python
#Categorizacion de los pacientes segun su peso
def clasificar_IMC(BMI):
  if BMI <  18.5:
    return "El paciente tiene bajo peso"
  elif BMI > 18.5 and BMI < 24.9:
    return "El paciente tiene un peso normal"
  elif BMI > 25 and BMI < 29.9:
    return "El paciente tiene sobrepeso"
  else:
    return "El paciente padece de obesidad"
```

```python
dpv["Categoria_IMC"] = dpv["BMI"].apply(clasificar_IMC)
#.apply() es un método de pandas que permite aplicar una función a cada elemento de una Serie (columna) o a cada fila de un DataFrame.
# Es como un for automático que recorre cada valor y le aplica una función.
#Sin apply seria necesario crear una lista vacia a la cual se le agreguen cada uno de los retornos de la funcion categoria_IMC para posteriormente ser agregada al dataset
```
